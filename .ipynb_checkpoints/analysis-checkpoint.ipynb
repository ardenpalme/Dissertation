```python
import yaml
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np
import os
```

## Helpers

```python
def get_l_perron_vec(arr_str):
    arr_str_clean = arr_str.strip('[]')
    return np.array([float(x) for x in arr_str_clean.split()])
```

```python
def purge_runs(df, run_ids, ledger_fname):
    df.drop(index=run_ids, inplace=True)
    for run_id in run_ids:
        fname = f"images/{run_id}.png"
        if os.path.exists(fname):
            os.remove(fname)
        else:
            print(f"File {fname} not found")
    df.to_csv(ledger_fname)
```

### 1. Impact of Graph Topology on Convergence and Generalization

* sign-flip attack
* training over all attacks
* $n=20$, $b=3$ for each topology
* stationary distribution is not uniform

```python
df1 = df.loc[['k-KDcnk530AmI0eslc-zJ', '6mfQOh1xcgLYNe8V0cuDQ', 'GgnvdtfLjPlVSzSUfeFix', 'Ywf6_2m4_gig66jorndcj']]
```

```python
fig, ax = plt.subplots(1,2, figsize=(12,4))
sns.barplot(
    data=df1.melt(
        id_vars=['graph_type'],
        value_vars=T0_perf_cols,
        var_name='Metric',
        value_name='Test Accuracy'
    ), 
    x='graph_type', 
    y='Test Accuracy', 
    hue='Metric',
    palette='muted6',
    width=0.5,
    ax=ax[0]
)
sns.barplot(
    data=df1.melt(
        id_vars=['graph_type'],
        value_vars=T3_perf_cols,
        var_name='Metric',
        value_name='Test Accuracy'
    ),
    x='graph_type', 
    y='Test Accuracy', 
    hue='Metric',
    palette='muted6',
    width=0.5,
    ax=ax[1]
)
ax[0].set_title("Byzantine-Free (T0) Performance")
ax[1].set_title("Byzantine (T3) Performance")
ax[0].set_xlabel("Communication Topology")
ax[1].set_xlabel("Communication Topology")
ax[0].legend(loc='lower right')
ax[1].legend(loc='lower left')

for a in [ax[0], ax[1]]:
    a.tick_params(axis='x', rotation=45)

plt.show()
```

```python
fig, ax = plt.subplots(1,3, figsize=(14,4))
sns.barplot(
    data=df1,
    x='graph_type', 
    y='byz_cons_floor', 
    width=0.5,
    ax=ax[0]
)
sns.barplot(
    data=df1,
    x='graph_type', 
    y='epsilon', 
    width=0.5,
    ax=ax[1]
)
# dB_max drives the byzantine consensus floor 
# this is the only value for consensus that is directly affected by the graph topology
sns.barplot(
    data=df1,
    x='graph_type', 
    y='dB_max', 
    width=0.5,
    ax=ax[2]
)

ax[0].set_title("1. Byzantine Consensus Noise")
ax[1].set_title("2. On-Avg Stability Upper Bound")
ax[2].set_title("3. Max Total Byz. Neighbor Weight")

for a in [ax[0], ax[1], ax[2]]:
    a.tick_params(axis='x', rotation=45)
fig.subplots_adjust(wspace=0.3)
plt.show()
```
