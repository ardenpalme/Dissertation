```python
import math
import threading
import time
from sklearn.datasets import fetch_openml
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler, RobustScaler, MinMaxScaler
from sklearn.decomposition import PCA
from scipy.special import softmax, log_softmax
from sklearn.model_selection import train_test_split, cross_validate, GridSearchCV
from sklearn.metrics import mean_squared_error, r2_score, accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
```

```python
global_seed = 0
global_rng = np.random.default_rng(global_seed)
```

```python
mnist = fetch_openml('mnist_784',parser = 'auto')
X = mnist.data
y = mnist.target.astype(int).to_numpy()
C = np.unique(y).shape[0]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=global_seed)

pipe = make_pipeline(
    MinMaxScaler(),
    PCA(n_components=0.93, svd_solver="full", random_state=0),
)

X_pca = pipe.fit_transform(X_train)
X_aug = np.hstack([X_pca, np.ones((X_pca.shape[0], 1))])
```

```python
def dirichlet_partition(labels, n_clients, alpha):
    labels = np.asarray(labels)
    client_idx = [[] for _ in range(n_clients)]

    for c in np.unique(labels):
        idx_c = np.where(labels == c)[0]
        global_rng.shuffle(idx_c)
        # proportion of class c assigned to each client
        proportions = global_rng.dirichlet(alpha * np.ones(n_clients))
        # split points (cumulative, dropping the last)
        splits = (np.cumsum(proportions)[:-1] * len(idx_c)).astype(int)
        for client, part in enumerate(np.split(idx_c, splits)):
            client_idx[client].extend(part.tolist())

    return [np.array(ci) for ci in client_idx]
```

```python
num_nodes = 20
batch_sz = 100
K = 100
mu = 0.2
lr = 0.3
b = 2
iter_sample_sz = 30 #math.floor(K/2)
```

```python
# the iterations at which we sample the system for trainin the logistic regression model
sampled_iters = global_rng.choice(K, size=iter_sample_sz, replace=False)
```

```python
B = global_rng.integers(0,num_nodes,size=b)
H = np.array(list(set(np.arange(num_nodes)) - set(B)))
```

```python
dp = dirichlet_partition(y_train, num_nodes, alpha=0.5)
max_agent_rows = np.max([dp[i].shape[0] for i in range(num_nodes)])
```

```python
d = max(4, b + 1)
while True:
    G = nx.random_regular_graph(d, num_nodes)
    if nx.node_connectivity(G) > b:
        break
```

```python
# Metropolis-Hasting weights
for i in range(num_nodes):
    for j in G.adj[i]:
        if(i != j):
            w_ij = 1/(1+max(G.degree[i], G.degree[j]))
            G.add_edge(i, j, weight= w_ij)

for i in range(num_nodes):
    out_weight_sum = np.sum(np.array([val['weight'] for key,val in G.adj[i].items()]))
    G.add_edge(i,i, weight = 1-out_weight_sum)

# assert nx.is_connected(G.subgraph(H)) and nx.node_connectivity(G) > b
```

```python
W = nx.to_numpy_array(G)
```

```python
def sgd_grad(X, y, theta, mu, batch_sz, rng):
    """Stochastic gradient of (1/n)Σ l_i + (mu/2)||theta||^2 on a uniform mini-batch.
       X:(n,d) bias-augmented, theta:(d,K), y:(n,) int, batch:int, rng:np.random.Generator."""
    n, K = X.shape[0], theta.shape[1]
    idx = rng.integers(0, n, size=batch_sz)  # uniform sample
    Xb, yb = X[idx], y[idx]
    Z  = Xb @ theta
    Yb = np.eye(K)[yb]
    loss = -(Yb * log_softmax(Z, axis=1)).sum() / batch_sz + 0.5 * mu * np.linalg.norm(theta)**2
    grad = Xb.T @ (softmax(Z, axis=1) - Yb) / batch_sz + mu * theta
    return loss, grad

def predict(X, theta):
    return (X @ theta).argmax(axis=1)     # argmax of logits == argmax of softmax
```

```python
barrier = threading.Barrier(num_nodes)   # optional: action=... runs once per release
models = np.zeros((num_nodes, X_pca.shape[1] + 1, C))     # x^k
int_models = np.zeros((num_nodes, X_pca.shape[1] + 1, C)) # x^{k+1}
```

```python
sampled_models = np.zeros((num_nodes, iter_sample_sz, X_pca.shape[1] + 1, C))
sampled_byz_labels = np.zeros((num_nodes, iter_sample_sz, 1))
```

```python
def worker(i, tar_node, tar_loss):
    sampled_iter_cnt = 0
    Xl, yl = X_aug[dp[i]], y_train[dp[i]]
    rng = np.random.default_rng(global_seed + i) # distinct stream per node
    for k in range(K):
        loss, g = sgd_grad(Xl, yl, models[i], mu, batch_sz, rng)
        int_models[i] = models[i] - lr * g       # local step
        barrier.wait()                           # (1) everyone wrote int_models[*]

        # if(i == tar_node):
        #     if(k in sampled_iters):
        #         print(f"iter {k}, idx {sampled_iter_cnt}")
        
        if(k in sampled_iters):
            sampled_models[i, sampled_iter_cnt] = models[i]
            sampled_byz_labels[i, sampled_iter_cnt] = 0
            sampled_iter_cnt = sampled_iter_cnt + 1

        models[i] = sum(W[i, j] * int_models[j] for j in G.neighbors(i))
            
        if(i == tar_node):# and k % 2 == 0):
            tar_loss.append(loss)
        barrier.wait()                           # (2) everyone finished mixing
```

```python
def byz_label_flip(i, tar_node, tar_loss):
    sampled_iter_cnt = 0
    Xl, yl = X_aug[dp[i]], y_train[dp[i]]
    rng = np.random.default_rng(global_seed + i) # distinct stream per node
    yl_perm = rng.permutation(yl)
    for k in range(K):
        loss, g = sgd_grad(Xl, yl_perm, models[i], mu, batch_sz, rng)
        int_models[i] = models[i] - lr * g             # local step
        barrier.wait()                                 # (1) everyone wrote half[*]

        # if(i == B[0]):
        #     if(k in sampled_iters):
        #         print(f"iter {k}, idx {sampled_iter_cnt}")
        
        if(k in sampled_iters):
            sampled_models[i, sampled_iter_cnt] = models[i]
            sampled_byz_labels[i, sampled_iter_cnt] = 1
            sampled_iter_cnt = sampled_iter_cnt + 1

        models[i] = sum(W[i, j] * int_models[j] for j in G.neighbors(i))
            
        barrier.wait()                           # (2) everyone finished mixing
```

```python
tar_loss = list()
tar_node = global_rng.integers(0,num_nodes)
```

```python
normal_threads = [threading.Thread(target=worker, args=(i, tar_node, tar_loss,)) for i in H]
byzantine_threads = [threading.Thread(target=byz_label_flip, args=(i, tar_node, tar_loss,)) for i in B]
threads = normal_threads + byzantine_threads

for thread in threads:
    thread.start()
    
for thread in threads:
    thread.join()
```

```python
def ce_loss(logits, y_true):
    """Mean cross-entropy. logits:(n,K), y_true:(n,) int labels."""
    logp = log_softmax(logits, axis=1)        # stable log-probabilities
    return -logp[np.arange(len(y_true)), y_true].mean()
    
X_pca_test = pipe.transform(X_test)
X_aug_test = np.hstack([X_pca_test, np.ones((X_pca_test.shape[0], 1))])

logits = X_aug_test @ models[tar_node]
print("test CE :", ce_loss(logits, y_test))
print("test acc:", (logits.argmax(1) == y_test).mean())
```

## Training Byzantine Classifier

### Pre-processing

```python
class MatrixSummaryFeatures(BaseEstimator, TransformerMixin):
    def __init__(self, outlier_threshold=3.0):
        self.outlier_threshold = outlier_threshold

    def fit(self, X, y=None):
        # X is expected to be (m, d, C)
        # No fitting required for these sample-wise statistics, but we keep the method
        return self

    def transform(self, X):
        # X: (m, d, C)
        # If someone passes a single matrix (d, C), reshape it to (1, d, C)
        if X.ndim == 2:
            X = X.reshape(1, X.shape[0], X.shape[1])
        
        m, d, C = X.shape
        eps = 1e-12

        # ------------------------------------------------------------
        # 1. Column norms (norm over the 'd' features for each class)
        # ------------------------------------------------------------
        col_norms = np.linalg.norm(X, axis=1)          # shape: (m, C)
        median_norm = np.median(col_norms, axis=1, keepdims=True)  # (m, 1)
        norm_ratios = col_norms / (median_norm + eps)   # (m, C)

        # ------------------------------------------------------------
        # 2. Cosine angle to median direction
        # ------------------------------------------------------------
        median_dir = np.median(X, axis=2)               # (m, d)  [median over classes]
        median_dir_norm = np.linalg.norm(median_dir, axis=1, keepdims=True)  # (m, 1)
        median_dir_unit = median_dir / (median_dir_norm + eps)  # (m, d)
        
        # Dot product of each column (d-dim) with the median direction
        # X: (m, d, C) , median_dir_unit: (m, d, 1) -> multiply -> (m, d, C), sum over d -> (m, C)
        cos_angles = np.sum(X * median_dir_unit[:, :, np.newaxis], axis=1) / (col_norms + eps)  # (m, C)

        # ------------------------------------------------------------
        # 3. Coordinate-wise outlier fraction
        # ------------------------------------------------------------
        # Row medians: median over classes (axis=2) for each feature 'd'
        row_medians = np.median(X, axis=2, keepdims=True)  # (m, d, 1)
        row_mads = np.median(np.abs(X - row_medians), axis=2, keepdims=True)  # (m, d, 1)
        
        # Modified z-score (using 0.6745 as scale constant for MAD)
        z_scores = 0.6745 * (X - row_medians) / (row_mads + eps)  # (m, d, C)
        outlier_mask = np.abs(z_scores) > self.outlier_threshold   # (m, d, C)
        
        # Fraction of outliers per feature 'd' -> then summarize across 'd'
        outlier_fraction_per_feature = np.mean(outlier_mask, axis=2)  # (m, d)
        outlier_summary = np.stack([
            np.mean(outlier_fraction_per_feature, axis=1),  # mean fraction (m,)
            np.max(outlier_fraction_per_feature, axis=1)    # max fraction (m,)
        ], axis=1)  # (m, 2)

        # ------------------------------------------------------------
        # 4. Combine all features
        # ------------------------------------------------------------
        features = np.concatenate([
            norm_ratios,        # (m, C)
            cos_angles,         # (m, C)
            outlier_summary     # (m, 2)
        ], axis=1)              # (m, 2*C + 2)
        
        return features
```

```python
# Assume C = 10 for example
C = 10

# Define column groups based on the new output (m, 2*C + 2)
norm_cols = list(range(0, C))                     # first C columns
angle_cols = list(range(C, 2*C))                 # next C columns
outlier_cols = list(range(2*C, 2*C + 2))         # last 2 columns

preprocessor = ColumnTransformer([
    ('norm_scaler', StandardScaler(), norm_cols),
    ('angle_scaler', RobustScaler(), angle_cols),  # robust scaler handles outliers in angles
    ('outlier_scaler', StandardScaler(), outlier_cols)
])

pre_pipeline = make_pipeline(
    MatrixSummaryFeatures(outlier_threshold=3.0),
    preprocessor,
    PCA(n_components=0.93, svd_solver="full", random_state=0),
)
```

## R-DSGD

```python
# system parameters
tau_log = 0.5
```

```python
def compute_tau(W, H, lr, gamma_C, X_aug, c_safety=3.0):
    h = len(H)
    Wh = W[np.ix_(H, H)]

    # Expected effective honest mixing matrix (Def. 4): off-diagonals
    # survive w.p. (1 - gamma_C); dropped/Byzantine mass returns to self-loop
    Wbar = (1.0 - gamma_C) * Wh
    np.fill_diagonal(Wbar, 0.0)
    np.fill_diagonal(Wbar, 1.0 - Wbar.sum(axis=1))

    # Stationary distribution pi (left Perron vector, Lemma 1)
    evals, evecs = np.linalg.eig(Wbar.T)
    pi = np.real(evecs[:, np.argmax(np.real(evals))])
    pi = np.abs(pi) / np.abs(pi).sum()
    pi_min, pi_max = pi.min(), pi.max()

    # Contraction factor on the disagreement subspace
    rho = np.linalg.norm(Wbar - np.outer(np.ones(h), pi), 2)

    # Mask-fluctuation constant (Lemma 4)
    w_max = (Wh - np.diag(np.diag(Wh))).max()
    nu2 = 2.0 * (pi_max / pi_min) * w_max * gamma_C * (1.0 - gamma_C)

    # Gradient bound for unregularized softmax LR (Frobenius norm on theta)
    R = np.linalg.norm(X_aug, axis=1).max()
    G = np.sqrt(2.0) * R

    gap = (1.0 - rho)**2 - nu2
    print(gap)
    assert gap > 0, "gamma_C too large for this graph's spectral gap"

    C_inf = (lr * G)**2 / gap                      # steady-state consensus error
    tau = 2.0 * c_safety * np.sqrt(C_inf / pi_min) + lr * G
    return tau
```

```python
def proj_tau(x_i, x_j, tau):
    d = x_j - x_i
    nd = np.linalg.norm(d)                # Frobenius for (d+1, C) matrices
    return x_i if nd == 0 else x_i + min(1.0, tau / nd) * d
```

```python
tau_proj = compute_tau(W, H, lr, fpr, X_aug) # gamma_C = fpr
```

```python
def worker_RDSGD(i, tar_node, tar_loss, pre_pipe, best_est, tau_log, tau_proj):
    sampled_iter_cnt = 0
    Xl, yl = X_aug[dp[i]], y_train[dp[i]]
    rng = np.random.default_rng(global_seed + i) # distinct stream per node
    for k in range(K):
        loss, g = sgd_grad(Xl, yl, models[i], mu, batch_sz, rng)
        int_models[i] = models[i] - lr * g       # local step
        barrier.wait()                           # (1) everyone wrote int_models[*]

        nbors = np.array([j for j in G.neighbors(i)])
        nbor_int_models_raw = np.array([int_models[j] for j in nbors])
        nbor_int_models = pre_pipe.transform(nbor_int_models_raw)
        nbor_byz_filt = best_est.predict_proba(nbor_int_models)[:,1] > tau_log
        
        if(nbor_byz_filt.shape != nbors.shape):
            print(f"nbor_byz_filt: {nbor_byz_filt.shape}, nbor : {nbors.shape}")
            print(nbors)
            print(best_est.predict_proba(nbor_int_models))
        assert(nbor_byz_filt.shape == nbors.shape)

        pred_byz_nbors = nbors[nbor_byz_filt] 
        for j in pred_byz_nbors:
            W[i,j] = 0
        W[i,i] = 1 - sum(W[i,j] for j in nbors)

        
        
        models[i] = sum(W[i, j] * proj_tau(models[i], int_models[j], tau_proj)
                        for j in list(G.neighbors(i)) + [i])

        if(i == tar_node and k % 2 == 0):
            tar_loss.append(loss)
        barrier.wait()                           # (2) everyone finished mixing
```

```python
# Byzantine node i doesn't know the identity of the other Byzantine nodes
def byz_label_flip_T3(i):
    sampled_iter_cnt = 0
    Xl, yl = X_aug[dp[i]], y_train[dp[i]]
    rng = np.random.default_rng(global_seed + i) # distinct stream per node
    yl_perm = rng.permutation(yl)
    for k in range(K):
        loss, g = sgd_grad(Xl, yl_perm, models[i], mu, batch_sz, rng)
        int_models[i] = models[i] - lr * g             # local step
        barrier.wait()                                 # (1) everyone wrote half[*]

        if(k in sampled_iters):
            sampled_models[i, sampled_iter_cnt] = models[i]
            sampled_byz_labels[i, sampled_iter_cnt] = 1
            sampled_iter_cnt = sampled_iter_cnt + 1

        models[i] = sum(W[i, j] * int_models[j] for j in G.neighbors(i))
            
        barrier.wait()                           # (2) everyone finished mixing
```

```python
honest_agent_threads = [threading.Thread(target=worker_RDSGD, args=(i, tar_node, tar_loss, pre_pipeline, grid.best_estimator_, tau_log, tau_proj,)) for i in H]
byz_agent_threads = [threading.Thread(target=byz_label_flip_T3, args=(i,)) for i in B]
RDSGD_threads = honest_agent_threads + byz_agent_threads

for t in RDSGD_threads:
    t.start()
    
for t in RDSGD_threads:
    t.join()
```
