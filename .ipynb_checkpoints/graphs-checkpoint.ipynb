```python
import networkx as nx
import numpy as np
import cvxpy as cp
from scipy.linalg import eigh
```

```python
global_seed = 0
global_rng = np.random.default_rng(global_seed)
```

```python
def create_graph(num_nodes, b, H, deg=6, seed=global_seed):
    rng = np.random.default_rng(seed)
    while True:
        graph_seed = int(rng.integers(2**31))
        G = nx.random_regular_graph(max(deg, b + 1), num_nodes, seed=graph_seed)
        if nx.node_connectivity(G) > b and nx.is_connected(G.subgraph(H)):
            break
            
    # Metropolis-Hasting weights
    for i in range(num_nodes):
        for j in G.adj[i]:
            if(i != j):
                w_ij = 1/(1+max(G.degree[i], G.degree[j]))
                G.add_edge(i, j, weight= w_ij)
    
    for i in range(num_nodes):
        out_weight_sum = np.sum(np.array([val['weight'] for key,val in G.adj[i].items()]))
        G.add_edge(i,i, weight = 1-out_weight_sum)

    W = nx.to_numpy_array(G)
    
    assert bool(np.sum(W.T @ np.ones(num_nodes)) == num_nodes) and bool(np.sum(np.ones(num_nodes).T @ W) == num_nodes)
    
    return G, W
```

```python
num_nodes = 20
b=2
B = global_rng.choice(num_nodes, size=b, replace=False)
H = np.array(list(set(np.arange(num_nodes)) - set(B)))
G,W = create_graph(num_nodes,b,H)
```

```python
r"""
Numerical solver for

    (P)   min_{W in S_n}  D_b(W)   s.t.   nu_b(W) >= nu,   ||w_{i,-i}||_0 <= d  for all i

    S_n    = { W = W^T >= 0 : W 1 = 1 }              (symmetric doubly stochastic)
    D_b(W) = max_i  phi_b(w_{i,-i})                  (sum of b largest off-diag entries of row i)
    nu_b(W)= min_{|B|=b} mu_2( L_{V\B}(W) )          (worst-case algebraic connectivity)

KEY POINT: no SDP solver is needed.  The only nonlinear constraint is
mu_2(L_H(W)) >= nu, and because

        z^T L_H(W) z = sum_{i<j in H} w_ij (z_i - z_j)^2        <-- LINEAR in W
        mu_2(L_H(W)) = min { z^T L_H(W) z : ||z||=1, z ⊥ 1 }

the constraint is an infinite family of LINEAR inequalities.  So (P) on a fixed
support is a semi-infinite LP, solved by Kelley cutting planes with HiGHS.

Variable ordering in the LP:  x = [ w (m) | u (2m) | s (n) | t (1) ]
Diagonal is eliminated:  w_ii = 1 - sum_{j~i} w_ij,  enforced by  sum_{j~i} w_ij <= 1.
"""

import numpy as np
from itertools import combinations
from scipy.optimize import linprog
from scipy.sparse import coo_matrix

# --------------------------------------------------------------------------
# graph helpers
# --------------------------------------------------------------------------

def circulant_regular(n, d):
    """Connected d-regular circulant C_n(1,...,d/2) (d even), plus antipodal if d odd."""
    assert (n * d) % 2 == 0
    E = set()
    for k in range(1, d // 2 + 1):
        for i in range(n):
            E.add(tuple(sorted((i, (i + k) % n))))
    if d % 2 == 1:                      # n must be even: add the perfect matching
        for i in range(n // 2):
            E.add(tuple(sorted((i, i + n // 2))))
    return sorted(E)


def random_regular(n, d, rng, swaps=None):
    """Random d-regular graph by double-edge swaps from a circulant (always succeeds)."""
    E = set(circulant_regular(n, d))
    swaps = swaps or 40 * len(E)
    for _ in range(swaps):
        El = list(E)
        (a, b_), (c, e) = El[rng.integers(len(El))], El[rng.integers(len(El))]
        if len({a, b_, c, e}) < 4:
            continue
        n1, n2 = tuple(sorted((a, c))), tuple(sorted((b_, e)))
        if n1 in E or n2 in E:
            continue
        E.discard((a, b_)); E.discard((c, e)); E.add(n1); E.add(n2)
        if not is_connected(n, sorted(E)):
            E.discard(n1); E.discard(n2); E.add((a, b_)); E.add((c, e))
    return sorted(E)


def is_connected(n, E):
    adj = {i: [] for i in range(n)}
    for i, j in E:
        adj[i].append(j); adj[j].append(i)
    seen, stack = {0}, [0]
    while stack:
        v = stack.pop()
        for w in adj[v]:
            if w not in seen:
                seen.add(w); stack.append(w)
    return len(seen) == n


def lam2_adjacency(n, E):
    A = np.zeros((n, n))
    for i, j in E:
        A[i, j] = A[j, i] = 1.0
    return np.sort(np.linalg.eigvalsh(A))[-2]

# --------------------------------------------------------------------------
# evaluation of the two functionals
# --------------------------------------------------------------------------

def full_W(n, E, w):
    W = np.zeros((n, n))
    for (i, j), val in zip(E, w):
        W[i, j] = W[j, i] = val
    np.fill_diagonal(W, 1.0 - W.sum(axis=1))
    return W


def D_b(W, b):
    """max_i (sum of b largest off-diagonal entries of row i)."""
    n = W.shape[0]
    off = W - np.diag(np.diag(W))
    return max(np.sort(off[i])[::-1][:b].sum() for i in range(n))


def mu2_fiedler(W, H, shift=5.0):
    """mu_2 and a cut vector z with ||z||=1 and z ⊥ 1, GUARANTEED.

    Deflation:  L' = L + (c/h) 11^T  has  L'1 = c*1  and  L' = L on 1-perp.
    Since mu_max(L) <= 2 < c under row-stochasticity, lambda_min(L') = mu_2(L)
    and its eigenvector is orthogonal to 1 -- even when L has a null space of
    dimension > 1 (disconnected induced graph), where plain eigh(L) may return
    a basis vector with a component along 1, producing an INVALID cut.
    """
    Wh = W[np.ix_(H, H)]
    Wh = Wh - np.diag(np.diag(Wh))
    L = np.diag(Wh.sum(axis=1)) - Wh
    h = len(H)
    Lp = L + (shift / h) * np.ones((h, h))
    vals, vecs = np.linalg.eigh(Lp)
    z = vecs[:, 0]
    z = z - z.mean()                    # numerical re-projection
    nz = np.linalg.norm(z)
    if nz < 1e-10:                      # degenerate; fall back to a split vector
        z = np.zeros(h); z[0] = 1.0; z -= z.mean(); nz = np.linalg.norm(z)
    z /= nz
    return float(z @ L @ z), z


def nu_b_exact(W, b, n):
    """Exhaustive min over |B| = b.  Returns (value, argmin B, Fiedler vector)."""
    best = (np.inf, None, None)
    for B in combinations(range(n), b):
        H = [v for v in range(n) if v not in B]
        m2, z = mu2_fiedler(W, H)
        if m2 < best[0]:
            best = (m2, B, z)
    return best

# --------------------------------------------------------------------------
# LP master problem with lazily generated cuts
# --------------------------------------------------------------------------

class Master:
    def __init__(self, n, E, b):
        self.n, self.E, self.b = n, E, b
        self.m = len(E)
        self.nbr = {i: [] for i in range(n)}
        for e, (i, j) in enumerate(E):
            self.nbr[i].append((e, j)); self.nbr[j].append((e, i))
        # directed pair index for the u-variables
        self.dir_idx = {}
        c = 0
        for i in range(n):
            for (e, j) in self.nbr[i]:
                self.dir_idx[(i, j)] = c; c += 1
        self.nu_dir = c                                   # = 2m
        self.nvar = self.m + self.nu_dir + n + 1
        self.iw, self.iu, self.iss = 0, self.m, self.m + self.nu_dir
        self.it = self.nvar - 1
        self.rows, self.rhs = [], []                      # A_ub x <= b_ub
        self._base_constraints()

    def _row(self):
        return np.zeros(self.nvar)

    def _base_constraints(self):
        n, b = self.n, self.b
        # (1)  b*s_i + sum_j u_ij - t <= 0
        for i in range(n):
            r = self._row()
            r[self.iss + i] = b
            for (e, j) in self.nbr[i]:
                r[self.iu + self.dir_idx[(i, j)]] = 1.0
            r[self.it] = -1.0
            self.rows.append(r); self.rhs.append(0.0)
        # (2)  w_ij - s_i - u_ij <= 0
        for i in range(n):
            for (e, j) in self.nbr[i]:
                r = self._row()
                r[self.iw + e] = 1.0
                r[self.iss + i] = -1.0
                r[self.iu + self.dir_idx[(i, j)]] = -1.0
                self.rows.append(r); self.rhs.append(0.0)
        # (3)  sum_{j~i} w_ij <= 1   (i.e. w_ii >= 0)
        for i in range(n):
            r = self._row()
            for (e, j) in self.nbr[i]:
                r[self.iw + e] = 1.0
            self.rows.append(r); self.rhs.append(1.0)

    def add_cut(self, B, z, nu):
        """ -sum_{(i,j) in E, i,j not in B} (z_i-z_j)^2 w_ij  <=  -nu """
        Bset = set(B)
        pos = {v: k for k, v in enumerate(v for v in range(self.n) if v not in Bset)}
        r = self._row()
        for e, (i, j) in enumerate(self.E):
            if i not in Bset and j not in Bset:
                r[self.iw + e] = -(z[pos[i]] - z[pos[j]]) ** 2
        self.rows.append(r); self.rhs.append(-nu)

    def solve(self):
        c = np.zeros(self.nvar); c[self.it] = 1.0
        bounds = ([(0, 1)] * self.m + [(0, None)] * self.nu_dir
                  + [(None, None)] * self.n + [(0, None)])
        A = coo_matrix(np.array(self.rows))
        res = linprog(c, A_ub=A, b_ub=np.array(self.rhs), bounds=bounds,
                      method="highs")
        if not res.success:
            raise RuntimeError(res.message)
        return res.x[self.iw:self.iw + self.m], res.fun

# --------------------------------------------------------------------------
# driver
# --------------------------------------------------------------------------

def solve_P_on_support(n, E, b, nu, max_iter=200, tol=1e-7, verbose=True):
    M = Master(n, E, b)
    hist = []
    for it in range(max_iter):
        w, lb = M.solve()
        W = full_W(n, E, w)
        val, Bstar, z = nu_b_exact(W, b, n)
        hist.append((lb, val))
        if verbose and (it < 5 or it % 10 == 0):
            print(f"  it {it:3d}   LP lower bnd {lb:.6f}   nu_b(W) {val:.6f}")
        if val >= nu - tol:
            if verbose:
                print(f"  converged at it {it}: LP bound {lb:.6f}, nu_b = {val:.6f}")
            return W, lb, hist
        # add the most violated cut, plus a few more for speed
        added = 0
        for B in combinations(range(n), b):
            H = [v for v in range(n) if v not in B]
            m2, zz = mu2_fiedler(W, H)
            if m2 < nu - tol:
                M.add_cut(B, zz, nu); added += 1
                if added >= 12:
                    break
    return W, lb, hist


def repair(W, b, n, nu):
    """Fact B rescaling W_t = (1-t)I + tW.  Feasible iff t*max_i(1-w_ii) <= 1."""
    val, _, _ = nu_b_exact(W, b, n)
    t = nu / val
    alpha = (1.0 - np.diag(W)).max()
    if t * alpha > 1 + 1e-9:
        return None, t
    Wt = (1 - t) * np.eye(n) + t * W
    return Wt, t


if __name__ == "__main__":
    rng = np.random.default_rng(0)
    n, b, d, nu = 12, 2, 6, 0.25
    print(f"instance: n={n}, b={b}, d={d}, nu={nu}   (need d >= b+1 = {b+1})")
    print(f"exhaustive |B| enumeration: C({n},{b}) = {len(list(combinations(range(n), b)))}\n")

    for name, E in [("circulant C_12(1,2,3)", circulant_regular(n, d)),
                    ("random 6-regular", random_regular(n, d, rng))]:
        lam2 = lam2_adjacency(n, E)
        closed = nu * b / (d - lam2 - b) if d - lam2 - b > 0 else np.inf
        print(f"--- {name}:  |E|={len(E)}, lambda_2(A)={lam2:.4f}, "
              f"gap d-lam2={d-lam2:.4f}")
        print(f"    Fact-A closed form  nu*b/(d-lam2-b) = {closed:.6f}")
        W, lb, hist = solve_P_on_support(n, E, b, nu)
        print(f"    LP-optimal D_b            = {D_b(W, b):.6f}")
        print(f"    realized nu_b             = {nu_b_exact(W, b, n)[0]:.6f}")
        off = W - np.diag(np.diag(W))
        nzp = off[off > 1e-9]
        print(f"    off-diag weights: min {nzp.min():.6f} max {nzp.max():.6f} "
              f"(uniform? {np.ptp(nzp) < 1e-6})")
        print(f"    self weights:     min {np.diag(W).min():.6f} "
              f"max {np.diag(W).max():.6f}")
        h = n - b
        lower = nu * b / d * (h - 1) / h
        print(f"    certificate lower bound nu*b/d*(h-1)/h = {lower:.6f}")
        print(f"    ratio achieved/lower = {D_b(W,b)/lower:.4f}\n")
```
