```python
%load_ext autoreload
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from pathlib import Path
from fastnanoid import generate
from sklearn.datasets import fetch_openml
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split, cross_validate, GridSearchCV, KFold
from datetime import datetime
```

```python
%autoreload 2
from graph_factory import GraphFactory, add_graph_plot
from proj_const_estimator import ProjConstEstimator
from utils import global_seed, rng, seed
from classifier import ByzClassifier
from system import SystemSimulator
```

### Global Dataset

```python
mnist = fetch_openml('mnist_784',parser = 'auto')
X = mnist.data
y = mnist.target.astype(int).to_numpy()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=global_seed)
```

```python
global_preproc = make_pipeline(
    MinMaxScaler(),
    PCA(n_components=0.93, svd_solver="full", random_state=global_seed, whiten=True),
    StandardScaler(),
)
```

```python
X_pca = global_preproc.fit_transform(X_train)
X_aug = np.hstack([X_pca, np.ones((X_pca.shape[0], 1))])
X_pca_test = global_preproc.transform(X_test)
X_aug_test = np.hstack([X_pca_test, np.ones((X_pca_test.shape[0], 1))])
```

```python
global_dataset = {
    'num_classes' : np.unique(y).shape[0],
    'X_train': X_aug,
    'y_train': y_train,
    'X_test': X_aug_test,
    'y_test': y_test
}
```

### Experiment Config

```python
%%capture
tar_dirichlet_alpha = 10
NUM_NODES = 20
config = {
    'alpha_init':0.5,         # 0.1, 0.3, 0.5, 1.0
    'alpha_decay': 0.1,       # 0.005, 0.01, 0.1, 1
    'graph_type':'geometric',
    'graph_weights':'MH_gen',
    'graph_args': {
        'geom_radius': 0.4,
        'dirichlet_alpha': tar_dirichlet_alpha,
        'MH_target_pi': rng('global', 'stationary-dist').dirichlet(np.full(NUM_NODES, tar_dirichlet_alpha))
    },
    'reg_param':0.2,
    'batch_sz':100, 
    'data_heterogeneity': 1,  # 0.1, 0.5, 10, 100
    'sys' : {
        'num_nodes': NUM_NODES,
        'b': 4,
        'K': 140,
        'atk_type':'sign_flip',
    },
    'train': {
        'gamma_C': 0.25,  
        'beta_C': 0.5,
        'train_atks': ['label_flip', 'sign_flip', 'gaussian'],
        'val_atks': ['label_flip', 'sign_flip', 'gaussian'],
        # 'num_nodes': 20,
        # 'b': 2,
        # 'K': 60,
    }
}
config['train']['num_nodes'] = config['sys']['num_nodes']
config['train']['b'] = config['sys']['b']
config['train']['K'] = config['sys']['K']

run_metrics = dict()
run_id = generate()
run_metrics['id'] = run_id
run_metrics['timestamp'] = pd.to_datetime(datetime.now())

isLogging = False
print_freq = 20

final_fig, final_ax = plt.subplots(2,3, figsize=(14,8))
```

### MISC

```python
gf = GraphFactory(config['train']['num_nodes'], config['train']['b'], seed('graph'))
```

```python
proj_const_estimator = ProjConstEstimator(config, global_dataset, rng('proj-const-estimator')) 
```

```python
proj_const_estimator.configure(gf,config)
```

```python
classifier.train_and_eval(X_C_train, y_C_train, X_C_test, y_C_test)
```

```python
classifier.plot(final_ax[1][1])
```

```python
C_opt_specs = classifier.clac_opt_operating_pt(0.7)
```

```python
C_best_est, C_pre_pipe = classifier.get_params()
```

```python
sys_sim = SystemSimulator(config, global_dataset, rng('ml-system-sim'))
```

```python
sys_sim.configure(gf, config, proj_const, C_opt_specs['fpr'], C_pre_pipe, C_best_est, C_opt_specs['tau'])
```

```python
sim_params = {
    'algorithms':['RDSGD', 'CooMed', 'SCC', 'TriMean', 'IOS'],
    'atk_type':config['sys']['atk_type'],
    'threat_model':'T0'
}
df = sys_sim.simulate(sim_params)
```

```python
%%capture
run_metrics['T0'] = dict()
iter_idx = np.arange(config['sys']['K'])
for alg in sim_params['algorithms']:
    final_ax[0,0].plot(iter_idx, df.loc[alg]['test_acc'], label=alg)

    run_metrics['T0'][alg] = dict()
    run_metrics['T0'][alg].update({
        'test_acc': df.loc[alg]['test_acc'].iloc[-1],
    })
final_ax[1,0].plot(iter_idx, df.loc['RDSGD']['Ck'])
final_ax[1,0].set_xlabel('Iteration k')
final_ax[0,0].set_ylabel('Test Accuracy')
final_ax[1,0].set_ylabel('Consensus (Ck)')
final_ax[0,0].set_title('Byzantine-Free (T0)')
final_ax[1,0].set_title('RDSGD Consensus Contraction')
```

```python
sim_params = {
    'algorithms':['RDSGD', 'CooMed', 'SCC', 'TriMean', 'IOS'],
    'atk_type':config['sys']['atk_type'],
    'threat_model':'T3'
}
df_t3 = sys_sim.simulate(sim_params)
```

```python
%%capture
run_metrics['T3'] = dict()
iter_idx = np.arange(config['sys']['K'])
for alg in sim_params['algorithms']:
    final_ax[0,1].plot(iter_idx, df_t3.loc[alg]['test_acc'], label=alg)
    final_ax[0,2].plot(iter_idx, df_t3.loc[alg]['train_loss'], label=alg)
    final_ax[1,2].plot(iter_idx, df_t3.loc[alg]['train_loss_pi'], label=alg)
    
    run_metrics['T3'][alg] = dict()
    run_metrics['T3'][alg].update({
        'test_acc': df_t3.loc[alg]['test_acc'].iloc[-1],
    })

final_ax[0,1].set_ylabel('Test Accuracy')
final_ax[0,1].set_title('Byzantine (T3)')

final_ax[0,2].set_ylabel('Training Loss')
final_ax[0,2].set_title('Byzantine (T3)')

final_ax[1,2].set_ylabel('Training Loss (pi-weighted)')
final_ax[1,2].set_title('Byzantine (T3)')
```
